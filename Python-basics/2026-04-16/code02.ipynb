```python
class A():
    x = int(input("Enter the value of x:-"))
    y = int(input("Enter the value of y:-"))
    z = x+y
    print(x)
    print(y)
    print(z)

print(A.y)    #address object
print(A.x)
print(A.z)    
print(A)

k = A()
print(k)
print(k.x)
print(k.y)          
print(k.z)

print(A().x)
print(A().y)        #accumate object
print(A().z)   

print(id(k))
print(id(A()))
```
