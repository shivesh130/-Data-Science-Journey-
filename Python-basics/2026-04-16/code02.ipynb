```python
class A():
    x = int(input("Enter the value of x:-"))
    y = int(input("Enter the value of y:-"))
    z = x+y
    print(x)
    print(y)
    print(z)

print(A.y)    #address object
print(A.x)
print(A.z)    
print(A)

k = A()
print(k)
print(k.x)
print(k.y)          
print(k.z)

print(A().x)
print(A().y)        #accumate object
print(A().z)   

print(id(k))
print(id(A()))
```

```python
x = 100
y = 200
x = x+y
y = x+y
print(x)
print(y)
def A():
    global x
    global y
    x = 1 
    y = 2             #we can over right the global variable in side local by using global keyword
    print(x)
    print(y)
A()
print(x)
print(y)  
```

```python
x = 100
y = 200
x = x+y
y = x+y
print(x)
print(y)
def A():
    nonlocal x             #we don't use nonlocal in function
    nonlocal y
    x = 1 
    y = 2             
    print(x)
    print(y)
A()
print(x)
print(y)  
```
