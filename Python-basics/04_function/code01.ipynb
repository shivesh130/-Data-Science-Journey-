```python
def name(*x):
    print(x[0])
    print(x[1])
    print(x[2])
name("shivesh","anjali","astha","shreya","purnima","beshti")  
print(name)
```

```python
def name(y,*x):
    print(y)
    print(x[1])
name("shivessh","ashish","mannan")    
```

```python
#positional argument
def A(x=2,y=3):
    print(x)
    print(y)
A(5,7)    
```

```python
#sequence conditional aargument
def A(x=4,y=6):
    print(x)
    print(y)
A(x = "shivesh",y = "anjali")    
```

```python
def A(x,y,*k):
    print(x)
    print(y)
    print(k)
A(x = "shivesh",y = "anjali",100,300,499)  
```

```python
def A(x,y,*k):
    print(x)
    print(y)
    print(k)
A(x = "shivesh",y = "anjali",k=(100,300,499))  
```
