# what is programming language?
* A programming language is a set of instructions used to communicate with a computer and perform specific tasks.

# why need of programming language?
* We need programming languages to give instructions to a computer in a structured way so that it can perform tasks and solve problems efficiently.

## Types of Programming Languages

* There are mainly two types of programming languages in the Information Technology world:

## Static Programming Language
* A language where the data type of a variable is defined at compile time.
Example: C, C++, java
##Dynamic Programming Language
* A language where the data type is decided at runtime.
Example: Python, JavaScript

# What is Python?

* Python was invented by Guido van Rossum in 1991.

Python is a high-level, interpreted, dynamically typed, and object-oriented programming language.
