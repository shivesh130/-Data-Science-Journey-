# filling values 
## 🌟 Concept
- In data cleaning, sometimes your dataset has missing values (NaN). Pandas provides methods like .fillna() to replace these missing values with something meaningful:

- A fixed value (e.g., 0, "Unknown")

- A statistical value (e.g., mean, median, mode)

- Forward fill (use previous value)

- Backward fill (use next value)

- This ensures your dataset is consistent and ready for analysis.

```python
df = pd.DataFrame({
        'Name': ['Amit', 'Riya', 'Karan', 'Neha'],
    'Age': [25, None, 30, None],
    'Score': [85, 90, None, 70]

})

print("orginial Dataframe:\n",df)

# fill missing age with a fixed value
df['Age'] = df['Age'].fillna(0)
print("-------------------------")
print(df)

#  Fill missing Score with the mean of the column

df['Score'] = df['Score'].fillna(df['Score'].mean())

print("---------------------")
print(df)

```

```python
# forward bill
import pandas as pd


df = pd.DataFrame({
    'A': [1, 2, None, 4],
    'B': [None, 5, 6, None]
})

# take previous value forward in column and axis 0 means work on column
df_fixed = df.ffill(axis=0)
print(df_fixed)

# take previous value forward in column and axis 0 means work on row
df_fixed = df.ffill(axis=1)
print(df_fixed)
```

```python
# backward bill
df = pd.DataFrame({
    'A': [1, 2, None, 4],
    'B': [None, 5, 6, None]
})

# take previous value backward in column and axis 0 means work on column
df_fixed = df.bfill(axis=0)
print(df_fixed)

# take previous value backward in column and axis 0 means work on row
df_fixed = df.bfill(axis=1)
print(df_fixed)
```

# Droping missing value
- Dropping missing values means removing rows or columns that contain NaN (null) values. This is useful when missing data cannot be meaningfully filled or when you want a clean dataset without gaps.


```python
import pandas as pd
import numpy as np

# Sample DataFrame
df = pd.DataFrame({
    'A': [1, 2, np.nan, 4],
    'B': [np.nan, 5, 6, np.nan],
    'C': [7, 8, 9, 10]
})

print("Original DataFrame:\n", df)
```

```python
# Drop Any Row with NaN (dropna()) 
# only take which have no nan value, all nan values rows are drop
df_dropna = df.dropna()
print(df_dropna)
```

```python
#Drop Rows Where All Values Are NaN
df_drop_all = df.dropna(how='all')
print(df_drop_all)
```

```python
# Drop Columns with NaN
df_drop_col = df.dropna(axis=1)
print(df_drop_col)

#Drop Columns Where All Values Are NaN

df_drop_col_all = df.dropna(axis=1, how='all')
print(df_drop_col_all)

```

```python
df_thresh = df.dropna(thresh=2)  # keep rows with at least 2 non-NaN
print(df_thresh)


#Drop NaN only if they appear in specific columns.
df_subset = df.dropna(subset=['A', 'B'])
print(df_subset)
```
