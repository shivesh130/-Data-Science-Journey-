### Data and time dtype

```python
import pandas as pd 
df = pd.DataFrame({
    'Emply':['A','B','C','D'],
    'join_date':['2023-09-04','2044-08-02','2034-05-03','2008-04-11']
})

print(df)
print(df.dtypes)
print("-----------converting to datatime-----------------")
df['join_date'] = pd.to_datetime(df['join_date'])
print(df.dtypes)

```

```python

# we can access the year, month,day separatly after converting 
print("----------------represent in year-----------------")
df['year'] = df['join_date'].dt.year
#print(df)
print("--------------represent in month----------------")
df['month'] = df['join_date'].dt.month
#print(df)
print("--------------represent in day----------------")
df['week'] = df['join_date'].dt.day_name()
df['week'] = df['join_date'].dt.day

print(df)
```

# Time indexing

```python
df = pd.DataFrame({
    'Emply':['A','B','C','D'],
    'join_date':['2023-09-04','2044-08-02','2034-05-03','2008-04-11'],
    'sales':[1500,2773,3000,6662]
})
df['join_date'] = pd.to_datetime(df['join_date'])
print(df)
print("------------------------------------")
df = df.set_index('join_date')
print(df)
print("------------------------------------")
print(df.loc['2044-08'])
print("------------------------------------")
print(df.iloc[1,'join_date'])
```

# resample
- it changes the frequency of your time series 
- turning daily sales into weekly or monthly totals

```python
week = df.resample('W').sum()
# print(week)
print("------------------------------------")
month = df.resample('ME').sum()
# print(month)
print("------------------------------------")
day = df.resample('D').sum()
print(day)
print("------------------------------------")
year = df.resample('YE').min()
print(year)
```

# Rolling windows
- A fixed size window that slieds across your data, one step at a time.only the n  values are considered at any point 
- syntxax:- df['column name'] = df['column name'].rolling(windows=n).agg_function

```python
df = pd.DataFrame({
    'Emply':['A','B','C','D'],
    'join_date':['2023-09-04','2044-08-02','2034-05-03','2008-04-11'],
    'sales':[1500,2773,3000,6662]
})

df['sales_roling'] = df['sales'].rolling(window=2).sum()
print(df)
print("------------------------------------------")
df['sales_rooling'] = df['sales'].rolling(window=4).mean()
print(df)
```

# Expanding 
- start with the first value then keep growing to include 
- everything seen so far
- syntxax:- df['column name'] = df['column name'].expanding(windows=n).agg_function

```python
df = pd.DataFrame({
    'Emply':['A','B','C','D'],
    'join_date':['2023-09-04','2044-08-02','2034-05-03','2008-04-11'],
    'sales':[1500,2773,3000,6662]
})

df['sum_sales'] = df['sales'].expanding().sum()
print(df)
print("------------------------------------------")
df['sales_mean'] = df['sales'].expanding().mean()
print(df)
```
