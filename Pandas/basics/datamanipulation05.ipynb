## Adding New column

```python
import pandas as pd
dic = {"name":['A','B','C','D','E'],
        'city':['delhi','kolkata','pune','mumbai','delhi'],
        "mark":[88,76,92,44,59],
       "age":[20,21,22,24,19]
       }
df = pd.DataFrame(dic)
df['dept'] = ['it','markt','finance','network',None]           #adding new column
print(df)
df['bonus'] = df['mark']-5   
print('------------------')     
print(df) 
```

```python
df.loc[2,'name'] = 66             #sigle value cahnge
print(df)
df.loc[0:2,'mark'] = 55            #multiple value cahnge
print(df)
```

```python
df.iloc[0,3]=77      #update single value
print(df)
print('------------------')     
df.iloc[0:2,3] = 100   #multiple value cahnge
print(df)
print('------------------')     
df.iloc[4,4] = 'xyz'
print(df)
```
