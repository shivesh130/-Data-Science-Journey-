## Adding New column

```python
import pandas as pd
dic = {"name":['A','B','C','D','E'],
        'city':['delhi','kolkata','pune','mumbai','delhi'],
        "mark":[88,76,92,44,59],
       "age":[20,21,22,24,19]
       }
df = pd.DataFrame(dic)
df['dept'] = ['it','markt','finance','network',None]           #adding new column
print(df)
df['bonus'] = df['mark']-5   
print('------------------')     
print(df) 
```

```python
df.loc[2,'name'] = 66             #sigle value cahnge
print(df)
df.loc[0:2,'mark'] = 55            #multiple value cahnge
print(df)
```

```python
df.iloc[0,3]=77      #update single value
print(df)
print('------------------')     
df.iloc[0:2,3] = 100   #multiple value cahnge
print(df)
print('------------------')     
df.iloc[4,4] = 'xyz'
print(df)
```

### apply function
* if we want to perfrom the same operation on every value in a column ,we use the apply function.

```python
import pandas as pd
dic = {"name":['A','B','C','D','E'],
        'city':['delhi','kolkata','pune','mumbai','delhi'],
        "mark":[88,76,92,44,59],
       "age":[20,21,22,24,19]
       }
df = pd.DataFrame(dic)
df['name'] = df['name'].apply(str.lower)
print(df['name'])
```

```python
df['mark'].apply(lambda mark:mark + 5)
df[mark]+5                                   #both are same
print(df)
```

### map()
- it is used to replace one value with another 
- it works only in a single column

```python
df = pd.DataFrame({
    "name": ["A", "B", "C", "D"],
    "score": [10, 20, 30, 40]
})
print(df)

# Replace names using dictionary
df["name"] = df["name"].map({"A": "Alpha", "B": "Beta", "C": "Gamma"})
print(df)
```

### sorting
* for single column in ascending order
- df.sort_value("column_name")
* for decending order 
- df.sort_values("xolumn_name", ascending=False)
* for multiple values
- df.sort_value(['column','column'])

```python
import pandas as pd
dic = {"name":['A','B','C','D','E'],
        'city':['delhi','kolkata','pune','mumbai','delhi'],
        "mark":[88,88,92,44,59],
       "age":[20,21,22,24,19]
       }
df = pd.DataFrame(dic)
# print(df.sort_values(['mark']))
print(df.sort_values('age',ascending=False))
```

```python
df.sort_index(ascending=True)              
```

### rank()
- it asssign eank to each value 

```python
df['rank'] = df['mark'].rank(ascending=False)
print(df)
df['rank'] = df['mark'].rank(ascending=False,method='dense')
print(df)
df['rank'] = df['mark'].rank(ascending=False,method='min')
print(df)
df['rank'] = df['mark'].rank(ascending=False,method='max')  #mean,first
print(df)
```
