# wide forma
- wide froma stores different on the same variable in a column each row usually reprent on entity. repeat mannegement at list multiple column this eaiser for human to read and useful for report and dashboard

```python
import pandas as pd

df = pd.DataFrame({
    'student':['A','B','B',"A"],
    'mark':[12,13,14,15],
    'sub':['math','sci','math','sci']
})

print(df)
print("-------------------------------")
wide_f = df.pivot(
    index='student',
    columns='sub',
    values='mark'
)
wide_f
```

# long format
- When to Use Long Format
- Analysis: Many statistical and plotting libraries (like Seaborn, R’s ggplot2) expect long format.

- Flexibility: Easier to group, aggregate, and filter by variable type.

- Machine‑friendly: One column holds values, another column tells you what those values represent.

```python
df = pd.DataFrame({
    'student':['A','B'],
    'math':[12,14],
    'sci':[15,13]
})
print(df)
print("-------------------------")


df_long = df.melt(id_vars='student', 
                  value_vars=['math','sci'],
                       var_name='sub', 
                       value_name='mark')

print(df_long)
```

# stacking 
- covert column into rows by moving column label into the index
- creating a hierarchical index

```python
df = pd.DataFrame({
    'student':['A','B'],
    'math':[12,14],
    'sci':[15,13]
})
print(df.stack())
print("----------------------")
print(df.unstack())
```
