# Handling mising value

- isnull() :- it return a dataframe of boolean value where True represent missing data 
- notnull() :-  it return a dataframe of boolean value where false represent missing data

```python
#isnull()
import pandas as pd

df = pd.read_csv(r"c:\Users\shive\OneDrive\Pictures\whatsapp\employees_details.csv")
print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
print((df.isnull().sum().sum()))
```

```python
print(df.notnull())
```

```python
print(df.notnull().sum())
```

```python
print(df.notnull().sum().sum())
```
