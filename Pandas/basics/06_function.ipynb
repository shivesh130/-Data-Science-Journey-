## concat
- it is used combine  dataframe vertically or horizontally

```python
import pandas as pd
dic = { "mark":[88,76,92,44,59]
       }
dic1= {"age":[20,21,22,24,19]}
df1 = pd.DataFrame(dic)
df2 = pd.DataFrame(dic1)
rsult2 = pd.concat([df1,df2],keys=['x','y','z'])
print("adding two or more dictonary and key are separated by dictonary")
print(rsult2)
rsult1 = pd.concat([df1,df2],ignore_index=True)
print("index ignore karega ")
print(rsult1)
rsult3 = pd.concat([df1,df2],ignore_index=True ,axis =1)
print("column mai add hoga")
print(rsult3)
rsult4 = pd.concat([df1,df2],ignore_index=True ,axis =0)
print("row mai add hoga")
print(rsult4)
```

## merge()
- Purpose: Combines rows from two DataFrames where values in key columns match.
- it work like the join in sql 

### Parameters:

- left, right: The two DataFrames you want to merge.

- on: Column(s) to join on (must exist in both DataFrames).

- how: Type of join:

- 'inner' → only matching rows (default).

- 'left' → all rows from left DataFrame + matches from right.

- 'right' → all rows from right DataFrame + matches from left.

- 'outer' → all rows from both, filling missing with NaN.

```python
dic = { "id":[3,4,9,6],
       "name":['A',"B","C","D"]
       }
dic1= {"id":[2,3,4,5],
       "name":['A',"B","C","D"]
       }
df1 = pd.DataFrame(dic)
df2 = pd.DataFrame(dic1)
print(pd.merge(df1,df2,on="id"))
print("--------------")
pd.merge(df1,df2,on="name")
# pd.merge(df1,df2,on="mark")
```

## inner merge()
- 

```python
pd.merge(df1,df2,on="id",how="inner")
```

```python
pd.merge(df1,df2,on="name",how="outer")
```

```python
pd.merge(df1,df2,on="id",how="right")
```

```python
pd.merge(df1,df2,on="id",how="left")
```
