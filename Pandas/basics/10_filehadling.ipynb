### Key Attributes:
- pd.read_csv("data.csv", sep=",", header=0, names=None, index_col=None, usecols=None,skiprows=None, nrows=None, na_values=None, dtype=None encoding="utf-8")

- sep → delimiter (, for CSV, \t for tab, ; etc.)

- header → row number to use as column names (0 = first row, None = no header)

- names → custom column names list

- index_col → column(s) to set as index

- usecols → select specific columns

- skiprows → skip certain rows at start

- nrows → read only limited number of rows

- na_values → specify strings to treat as NaN

- dtype → force column data types

- encoding → handle special characters (e.g., "utf-8", "latin1")
