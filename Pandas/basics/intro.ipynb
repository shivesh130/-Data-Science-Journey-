# Data structure pandas

```python
import pandas as pd
```

```python
arr = pd.Series(['a','b','c','d'])
arr.name="alphbate"
print(arr)
```

```python
arr =pd.Series([10,20,30,"40",50])
print(arr)
```

```python
# using dictionary
arr = pd.Series({"a":10,"b":20,"c":30})
print(arr)
```

```python
arr = pd.Series([10,20,30,40],index=["a","b","c","d"])
print(arr)
```

```python
# scalar
arr =pd.Series(10,index=["a","b","c","d"])
print(arr)
```

# manual indexing

```python
arr = pd.Series([10,20,30,40,50])
print(arr[1])
print(arr[3])
print(arr[4])
```

# slicing

```python
arr = pd.Series([10,20,30,40,50])
print(arr[1:4])
print(arr[0:-1])
```

# iloc and loc 

* iloc works as index
* loc works as labeling

```python
arr =pd.Series([10,20,30,90],index=["a","b","c","d"])    #labeling
print(arr.iloc[3])
print('------')
print(arr)
print('------')
print(arr.loc["b"])
```
