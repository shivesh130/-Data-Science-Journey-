### groupby     

- the groupby function use to split tha data into group base on one or more columns after grouping we can perfrome calculation of ecah group

```python
import pandas as pd
dic = { "id":[101,102,103,104],
       "name":['Aa',"Aa","Cc","Dd"]
       ,'age':[20,90,30,None]
       }
dic1= {"id":[2,3,4,5],
       "name":['A',"B","C","D"]
       }
df1 = pd.DataFrame(dic)
df2 = pd.DataFrame(dic1)
df1.groupby("name")

```

```python
dic = { "id":[101,102,103,104],
       "name":['Aa',"Aa","Cc","Dd"]
       ,'age':[20,90,30,None]
       }
df1 = pd.DataFrame(dic)
df1.groupby('name')['age'].sum()
```

```python
dic = { "id":[101,102,103,104],
       "name":['Aa',"Aa","Cc","Dd"]
       ,'age':[20,90,30,None]
       ,'marks':[45,55,65,75]
       }
df1 = pd.DataFrame(dic)
df1.groupby('name')['age'].agg(['sum','count'])
df1.groupby('name')['age'].agg(['sum','count','max'])
```

```python
import pandas as pd

dic = { 
    "id":[101,102,103,104],
    "name":['Aa',"Aa","Cc","Dd"],
    "age":[20,90,30,None],
    "marks":[45,55,65,75]
}
df1 = pd.DataFrame(dic)

result = df1.groupby('name').agg({
    'age':['sum','mean'],
    'marks':'count'
})
print(result)
```

### transform filter
- it performe a calculation each group return result with same number of original

```python

dic = { 
    "id":[101,102,103,104],
    "name":['Aa',"Aa","Cc","Dd"],
    "age":[20,90,30,None],
    "marks":[45,55,65,75]
}
df1 = pd.DataFrame(dic)
df1['avesalry']= df1.groupby('name')['age'].transform('mean')
print(df1)
```

### filter
- keep only the group that satisfy a condition 


```python
df1.groupby('name').filter(lambda x: x ['age'].mean()>25)
```
