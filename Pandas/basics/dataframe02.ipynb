# DataFrame 
## Key Features of Pandas DataFrame
* Structure: Two dimensions — rows and columns.

* Data Types: Can store integers, floats, strings, dates, categorical values, etc.

* Labels: Each row has an index; each column has a name (label).

* Flexibility: Columns can hold different data types.

* Integration: Built on top of NumPy, making numerical operations fast.

# From External Files

* CSV: pd.read_csv("file.csv")

* Excel: pd.read_excel("file.xlsx")

* SQL: pd.read_sql(query, connection)


```python
import pandas as pd
list =[[1,2,3,4],[5,6,7,8]]
df = pd.DataFrame(list,columns=['A','B','C','D'],dtype='int32')
print(df)
```

```python
dic = {"name":['A','B','C','D','e'],"rollno":[101,102,103,104,105],"score":[25,35,45,65,55]}
df  = pd.DataFrame(dic,index=['A','B','C','D','e'])
print(df)
print("----------------")
print(df['name'],'\n',df['rollno'])
```
