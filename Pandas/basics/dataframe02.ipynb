# DataFrame 
## Key Features of Pandas DataFrame
* Structure: Two dimensions — rows and columns.

* Data Types: Can store integers, floats, strings, dates, categorical values, etc.

* Labels: Each row has an index; each column has a name (label).

* Flexibility: Columns can hold different data types.

* Integration: Built on top of NumPy, making numerical operations fast.

# From External Files

* CSV: pd.read_csv("file.csv")

* Excel: pd.read_excel("file.xlsx")

* SQL: pd.read_sql(query, connection)


```python
import pandas as pd
list =[[1,2,3,4],[5,6,7,8]]
df = pd.DataFrame(list,columns=['A','B','C','D'],dtype='int32')
print(df)
```

```python
dic = {"name":['A','B','C','D','e'],
       "rollno":[101,102,103,104,105],
       "score":[25,35,45,65,55]
       }
df  = pd.DataFrame(dic,index=['A','B','C','D','e'])
print(df)
print("----------------")
print(df['name'],'\n',df['rollno'])
print('-------------')
print(df.iloc[2]["rollno"])
print(df.query("score < 35 & rollno < 104"))
```

```python
print(type(df["name"]))
```

```python
print(df[['name','rollno']])
```

```python
dic = {"name":['A','B','C','D','e'],"rollno":[101,102,103,104,105],"score":[25,35,45,65,55]}
df  = pd.DataFrame(dic,index=['A','B','C','D','e'])
print(df.loc['A',"rollno"])             #using 

```

```python
dic = {"name":['A','B','C','D','e'],
       "rollno":[101,102,103,104,105],
       "score":[25,35,45,65,55]}
df = pd.DataFrame(dic)
print(df)
print("--------------")
print(df.iloc[2]["rollno"])
print("--------------")
print(df.iloc[1,0])      #row, column
print(df.iloc[1,2])
print(df.iloc[1,1])
```

```python
# slicing
print(df.iloc[0:2])              #row
print(df.iloc[:,0:2])        #column
```

```python
#boolean filttering
print(df['score']>55)
print("-----------------")
print(df[df['score']>55])
```

```python
#basic operator
print(df)
print(df['score'].mean())
print(df.groupby('name')['score'].mean())         #groupby use for combine two column
```

```python
dic = {"name":['A','B','C','D','e'],
       "rollno":[101,102,103,104,105],
       "score":[25,35,45,65,55]}
df = pd.DataFrame(dic)
df['score']=df['score'].astype('str')             #type casting
print(df['score'])
```
