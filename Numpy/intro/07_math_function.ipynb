# math function

## axis based operation

```python
import numpy as np
a = np.array([[10,40,50],[20,30,40]]) 
x = np.sum(a[1],axis = 0)
print(x)
print(sum(a))
```

# rounding function
- the function in numpy round the element of an array to specific of element
1. Round()
2. floor()
3. ceil()

```python
a = np.array([12.33,56.4544,455.223,-224.2223])
print(np.round(a))
print(np.round(a,2))
print(np.floor(a))
print(np.ceil(a))

```

## logrithm function(logn,log())

```python
a = np.array([1,22.334,44,62.34])
print(np.log(a))
print(np.log2(a))
print(np.log10(a))
print(np.log(a).round(3))
```

## Trigonometric function

```python
a = np.array([10,20,30,40])
print(np.sin(a))
print(np.cos(a))
print(np.tan(a))
print(np.radians(a))
print(np.arcsin(a))
print(np.degrees(np.arcsin(a)))
print(np.degrees(a))
```
