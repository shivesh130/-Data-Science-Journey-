# histrogram
- A histogram is a way to visualize the distribution of data. It shows how often values fall into different ranges (called bins). Think  of it as grouping numbers into buckets and then drawing bars to represent how many values are in each bucket.

- Key Features
1. Bins: Intervals that divide the range of data (e.g., 0–10, 10–20, etc.).

2. Bars:  Height of each bar shows how many data points fall into that bin.

3. X-axis: Represents the data values (ranges).

4. Y-axis: Represents frequency (count of values in each bin).

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 random numbers from a normal distribution
data = np.random.normal(loc=50, scale=10, size=1000)

# Create histogram with 20 bins
plt.hist(data, bins=20, color='skyblue', edgecolor='black')

plt.title("Histogram Example")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```
