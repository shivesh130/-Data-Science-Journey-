# insorting data
## insert

```python
import numpy as np
x = np.array([10,20,30,40,50,66,7,0])
d = np.insert(x,4,4)
print(d)
```

## delete data

```python
x = np.array([10,20,30,40,50,66,7,0])
d = np.delete(x,4)
print(d)
```

## append

```python
x = np.array([10,20,30,40,50,66,7,0])
d = np.append(x,55)
c = np.append(x,(2,6,5,677,9))
print(d)
print(c)
```
