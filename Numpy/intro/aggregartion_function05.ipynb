## aggregation function

```python
import numpy as np

arr1 =np.array([1,2,3,4,5,6,7,8])
arr2 =np.array([10,20,30,40,50,60,70,80])

print(np.greater(arr2,arr1))
print(np.greater(arr1,arr2))
print("-------------------")
print(np.greater_equal(arr2,arr1))
print(np.greater_equal(arr1,arr2))
print("-------------------")
print(np.less(arr1,arr2))
print(np.less_equal(arr1,arr2))
print("-------------------")
print(np.equal(arr1,arr2))
print(np.not_equal(arr1,arr2))
print("-------------------")
print(np.subtract(arr1,arr2))
print(np.add(arr1,arr2))
print(np.multiply(arr1,arr2))
print(np.divide(arr1,arr2))
print(np.floor_divide(arr1,arr2))
print("-------------------")
# print(np.max(arr2,arr1))
# print(np.min(arr1,arr2))
print("-------------------")
print(np.square(arr1,arr2))
print("-------------------")
# print(np.sqrt(arr1,arr2))
print(np.abs(arr1,arr2))
print("-------------------")
# print(np.mean(arr1,arr2))
# print(np.median(arr1,arr2))
# print(np.std(arr1,arr2))
# print(np.var(arr1,arr2))
# print(np.mod(arr1,arr2))
# print(np.power(arr1,arr2))
```
