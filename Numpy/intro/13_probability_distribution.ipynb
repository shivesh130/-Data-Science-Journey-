## uniform distribution

```python
import numpy as np

m = np.random.uniform(1,10,2)
print(m)
```

## Normal distribution
- values ine dirtibution around a central mean forming a bell-shaped curve
- loc → mean (μ) of the distribution

- scale → standard deviation (σ)

- size → shape of the output array (number of samples)

```python
np.random.normal(loc=50,scale=10,size=10)
```

## Binomial distribution
- Key Features
- Trials (n): Number of experiments (e.g., flipping a coin 10 times).

- Probability (p): Chance of success in each trial (e.g., probability of heads = 0.5).

- Random variable (X): Number of successes out of n trials.

- Range: 𝑋 can take values from 0 up to n.

```python
import numpy as np

# Toss a coin 10 times, probability of heads = 0.5
x = np.random.binomial(n=10, p=0.5, size=5)
print(x)
```

## poisson distribution
-  Key Features
- Events are rare and independent (e.g., number of phone calls per hour, accidents per day).

- Defined by a single parameter: λ (lambda) = average rate of occurrence.

- Random variable 𝑋 = number of events in the interval.

- Possible values: 0,1,2,…(non-negative integers).

```python
import numpy as np

# Average 5 events per interval
x = np.random.poisson(lam=5, size=10)
print(x)
```

## Exponential distribution
- Key Features
- Defined by a single parameter: λ (lambda) = rate of occurrence.

- Random variable 𝑋 = time until the next event.

- Always non-negative (𝑋≥0).

- Memoryless property: probability of waiting longer doesn’t depend on how much time has already passed.

```python
import numpy as np

# Average waiting time = 5 units (scale=5)
samples = np.random.exponential(scale=5, size=10)
print(samples)
```
