## arthmethamatic operation

```python
import numpy as np
```

```python
arr1 =np.array([1,2,3,4,5,6,7,8])
arr2 =np.array([10,20,30,40,50,60,70,80])

print("addition",arr1+arr2)
print("subration",arr1-arr2)
print("division",arr1/arr2)
print("modules",arr1%arr2)
print("multiple",arr1*arr2)
print("power",arr1**arr2)
print("floor",arr1//arr2)
```
