## File handleing
* It is used when you want to store arrays permanently and use
* them later without recreating them .

```python
import numpy as np 
name = np.array(['shivesh','anjali','sumnat','soulmaya'])
age = np.array([19,20,30,40])
roll_no = np.array([1,2,3,4,5])
student = np.array(["shivesh",19,"shudip",99])
print(name)
print(age)
print(roll_no)
print(np.save("age.npy",age))
```

```python
np.save("age.npy",age)
print("\nage.npy file created successfully")

```

## save multiple files

```python
np.savez(
    "shivesh.npz",
    name = name,
    age = age,
    roll_no = roll_no
)
print("saved")
```

## load the save file

```python
stu = np.load("shivesh.npz")
print(stu)
```

## save as text file

```python
np.savetxt(
    "d.txt",
    name,
    fmt = "%s"

)
```

```python
np.savetxt(
    "save.txt"
    , age
    , fmt="%d"

)
```

##  Load The Text file

```python
text_data = np.loadtxt(
    "save.txt",
    dtype=int,
    )

print("\nmark.text file load successfully")

print(text_data)
```

```python
text_name = np.loadtxt(
    "d.txt",
    dtype=str)

print("\nname.text file load successfully")

print(text_name)
```

## Save as CSV file : comma separate values

```python
student_rec = np.array([
    [101,85],
    [102,86],
    [103,87]
])
np.savetxt(
    "student.csv",
    student_rec,
    delimiter= ",",
    fmt = "%d"
)
print("\nstudents.csv file save successfully")
```

## Load csv file

```python
csv_data = np.loadtxt(
    "student.csv",
    delimiter=",",
    dtype=int)
print("\nData loaded from CSV")
print(csv_data)
```

```python

student1 = np.array([
    [101,'xyz',85],
    [102,'pqr',56],
    [103,'abc',43],
    [104,'lmn',78]])
np.savetxt(
    "students1.csv",
    student1,
    delimiter=",",
    fmt="%s")
print("\nstudents.csv file save successful")
```
