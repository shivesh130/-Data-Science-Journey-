```python
numpy is short from of numerical python. it is a open source library that provdies support for large multi dimensional array 4 mattix,
 along with large collection of high -level math function to operate on there array, created by travis oliphant in 2005. 

numpy array is a homogenoues data structure. it was optimize  c language style storage where actual values are store in contigous.
```

```python
import numpy as np
```

```python
#1D
A = np.array([1,2,3,4,5])
print(A)
print(A.ndim)

#2D
B = np.array([["shivesh"],
              ["anjlai"]])
print(B)
print(B.ndim)

#3D
c = np.array([[[1,2,3,4],
               [5,6,7,8]]])
print(c)
print(c.ndim)
```

```python
d = np.zeros(4)
print(d)
f = np.ones(8)
print(f)
g = np.empty(3)
print(g)
```

```python
#arnage(start,stop,step)
d = np.arange(0,10)
print(d)
s = np.arange(1,10,2)
print(s)
```

```python
d = np.linspace(1,10,4)
print(d)
```

```python
d = np.array([[1,2,3],
             [4,5,6]])
print(d)
print(d.shape)
print(d.size)
print(d.ndim)
print(d.dtype)
print(d.itemsize)
print(d.nbytes)
print(d.dtype.type)
```

```python
d = np.array([[[1],
               [2],
               [3]],
               [[4],
                [5],
                [6]]],dtype=np.int16)
print(d.ndim)
print(d.dtype)
print(d.dtype.type)
```
