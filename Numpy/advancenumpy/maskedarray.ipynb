## masked array
* masked array allows you to hide or ignore certain values perfroming calculation.

```python
import numpy as np
import numpy.ma as ma

#ma - masked array madule inside numpy

data = np.array([3,-4,7,8,9])
masked = np.ma.masked_equal(data,7)
print(masked)
# if  thin are more than one negative values
data1 = np.array([3,-4,-7,8,9])

mask = np.ma.masked_where((data1 == -4)|(data1==-7),data1)
print(mask)
```
