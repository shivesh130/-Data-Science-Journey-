```python
import pandas as pd 
import matplotlib.pyplot as plt

dat = {
    'month':['jan','feb','mar','april'],
    'sales':[120,239,21,552],
    'profit':[20,30,40,45]
}
df = pd.DataFrame(dat)
print(df)

```

```python
df.plot(
    x = 'sales',
    y = 'profit',
    kind = 'hexbin'
)
plt.show()
```

```python
df.plot(
    x = 'month',
    y = 'sales',
    kind = 'line',
    marker = 'o',
    color = 'green'
)
plt.title('monthly sale')
plt.xlabel('month')
plt.ylabel('sales')
plt.grid(True)
plt.show()
```

```python
dat = {
    'month':[1,2,3,4],
    'sales':[120,239,21,552],
    'profit':[20,30,40,45]
}
df = pd.DataFrame(dat)
fig = plt.figure()
ax = fig.add_subplot(111,projection='3d')
ax.plot(
    df['month'],
    df['sales'],
    df['profit'],
    marker= 'o',
    color= 'red'
)
ax.set_xlabel('month')
ax.set_ylabel('sales')
ax.set_zlabel('profit')
plt.show()
```

```python
data = {
    'sales':[120,203,340,459,332]
}
df= pd.DataFrame(data)
df['sales'].plot(kind='hist',bins=5)
plt.title('sales distribution')
plt.xlabel('sales')
plt.ylabel('frequency')
plt.grid(True)
plt.show()
```
